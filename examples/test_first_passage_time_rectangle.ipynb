```python
import numpy as np
import matplotlib.pyplot as plt

Lx = 1.0
Ly = 1.0

pos0 = np.array([0.5, 0.5])

sigma = 1
drift = np.array([0.0, 0.0])
```

```python
n_particles = 10000
dt = 0.001


# exit times for x == 0
exit_times_0 = []
exit_positions_0 = []
# exit times for x == 1
exit_times_1 = []
exit_positions_1 = []
# exit times for y == 0
exit_times_2 = []
exit_positions_2 = []
# exit times for y == 1
exit_times_3 = []
exit_positions_3 = []

for i in range(0, n_particles):

    pos = pos0.copy()
    t = 0

    while True:
        pos += drift * dt + sigma*np.random.normal(0, 1, 2) * np.sqrt(dt)
        t += dt

        if(pos[0] < 0):
            exit_times_0.append(t)
            exit_positions_0.append(pos)
            break
        if(pos[0] > 1):
            exit_times_1.append(t)
            exit_positions_1.append(pos)
            break
        if(pos[1] < 0):
            exit_times_2.append(t)
            exit_positions_2.append(pos)
            break
        if(pos[1] > 1):
            exit_times_3.append(t)
            exit_positions_3.append(pos)
            break
```

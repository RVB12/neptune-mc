```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
n_particles = 10000

sigma = 1
drift = 0
# drift = 0.5
dt = 0.0001

times = np.zeros(n_particles, dtype=float)
positions = np.zeros(n_particles, dtype=float)

for i in range(0, n_particles):
    x = 0.5
    y = 0.5
    t = 0

    while True:
        x += drift*dt + sigma*np.random.normal(0, 1) * np.sqrt(dt)
        y += sigma*np.random.normal(0, 1) * np.sqrt(dt)
        t += dt
        if(x > 1):
            positions[i] = y
            times[i] = t
            break
        if(x < -1 or y < -1 or y > 1 or t > 100):
            break

print(times)
```

```python
# plot the histogram along with the theoretical solution

# theoretical solution, levy distribution
t = np.linspace(0, 100, 1000)
alpha = 1
p = alpha / (sigma * np.sqrt(2*np.pi*t**3)) * np.exp(-(alpha-drift)**2 / (2*sigma**2*t))

plt.hist(times, bins=1000, density=True, label='Simulation')
plt.plot(t, p, label='Theoretical solution')
plt.legend()
plt.show()
```

```python
plt.hist(positions, bins=1000, density=True, label='Simulation')

da = 0.5
db = 1.5

alpha = db / (db + da)
beta = da / (db + da)

print(alpha, beta)



analytical = 1 - 2 * np.random.beta(1/alpha, 1/beta, 10000)
plt.hist(analytical, bins=100, density=True, label='Analytical solution')
plt.legend()
plt.show()
```
